### 과제

```python
import wget, cv2
import numpy as np

wget.download('https://raw.githubusercontent.com/tysep16/CV26_1/refs/heads/main/snp.jpg', './img/')

image = cv2.imread('./img/snp.jpg')
median_filtered = cv2.medianBlur(image, 5)
full_image = np.hstack([image, median_filtered])

cv2.imshow('output', full_image)
cv2.waitKey(0)
cv2.destroyAllWindows()
cv2.waitKey(1)
```

```python
import cv2
import numpy as np

image = cv2.imread('./img/snp.jpg')
gaussian_filtered = cv2.GaussianBlur(image, (0, 0), 2)
full_image = np.hstack([image, gaussian_filtered])

cv2.imshow('output', full_image)
cv2.waitKey(0)
cv2.destroyAllWindows()
cv2.waitKey(1)
```
