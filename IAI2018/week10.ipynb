1. cv2.warpAffine() 함수를 이용하여 이동 변환해 보세요.
* X축으로 50, y축으로 50

```python
import cv2
import numpy as np

image = cv2.imread('./img/lenna.png')
tMatrix = np.array([[1, 0, 50], [0, 1, 50]], dtype=np.float32)
final = cv2.warpAffine(image, tMatrix, [0, 0])

cv2.imshow('final', final)
cv2.waitKey()
cv2.destroyAllWindows()
```

2. cv2.warpAffine() 함수를 이용하여 전단 변환해 보세요.
* x’ = x+0.5y 만큼, y’ = 0.5x + y 만큼 전단 변환

```python
import cv2
import numpy as np

image = cv2.imread('./img/lenna.png')
sMatrix = np.array([[1, 0.5, 0], [0.5, 1, 0]], dtype=np.float32)
h, w, _ = image.shape
final = cv2.warpAffine(image, sMatrix, [w+h//2, h+w//2])

cv2.imshow('final', final)
cv2.waitKey()
cv2.destroyAllWindows()
```
