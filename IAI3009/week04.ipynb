## 머신러닝응용 4주차 과제

1. vscode를 이용하여 임베디드 시스템에 접속한 후, 아래 주소의 데이터를 읽어와 데이터프레임을 생성하세요.

```python
import pandas as pd
df = pd.read_csv('https://raw.githubusercontent.com/tysep16/26_02_ML/refs/heads/main/SeoulBike.csv')
df
```

2. `.isnull()`과 `sum()`을 이용하여 결측치 개수를 확인하세요.

```python
df.isnull().sum()
```

3. 컬럼 이름(라벨)이 아래 결과와 같도록 수정하세요.

```python
df.columns = ['date', 'rented', 'hour', 'temp', 'humidity', 'wind', 'visibility', 'dew', 'solar', 'rain', 'snow', 'season', 'holiday', 'functioning']
df
```

4. 아래 결과와 같게 seaborn 패키지를 이용하여 상관계수를 heat map으로 시각화 화세요.

```python
import seaborn as sns
import matplotlib.pyplot as plt

corr = df.corr(numeric_only=True)
sns.heatmap(corr, annot=True, fmt='.2f', cmap='coolwarm')
plt.show()
```

5. 아래 결과와 같이 데이터 프레임 df의 컬럼 라벨을 출력해보세요.

```python
hour_mean = df.groupby('hour')['rented'].mean()
plt.bar(hour_mean.index, hour_mean, color='skyblue')
plt.xlabel('hour')
plt.ylabel('mean rented')
plt.show()
```
