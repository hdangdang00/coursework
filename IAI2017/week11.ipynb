### 실습 과제

1. titanic.csv 'Pclass', 'SibSp', 'Parch', 'Fare' 열을 x_input 데이터로 'Survived' 열은 y_output 데이터로 나누고 train/test 데이터 세트를 생성하세요.

```python
import pandas as pd
from sklearn.model_selection import train_test_split

df = pd.read_csv('https://raw.githubusercontent.com/tysep16/ML26_01/refs/heads/main/titanic.csv')
x_input = df[['Pclass', 'SibSp', 'Parch', 'Fare']]
y_output = df['Survived']
x_train, x_test, y_train, y_test = train_test_split(x_input, y_output, random_state=42)
```

2. StandardScaler 클래스를 이용하여 표준화 하세요.

```python
from sklearn.preprocessing import StandardScaler

ss = StandardScaler()
ss.fit(x_train)
x_train_scaled = ss.transform(x_train)
x_test_scaled = ss.transform(x_test)
```

3. 랜덤포레스트, XGBoost, SVM을 이용하여 각각 모델을 생성하고 평가하세요.

* 2번에서 생성한 표준화 데이터 사용 X

```python
# Random Forest
from sklearn.ensemble import RandomForestClassifier

rf = RandomForestClassifier(n_jobs=-1)
rf.fit(x_train, y_train)
print('Random Forest score:', rf.score(x_test, y_test))
```

```python
#  XGBoost
from xgboost import XGBClassifier

xgb = XGBClassifier()
xgb.fit(x_train, y_train)
print('XGBoost score:', xgb.score(x_test, y_test))
```

```python
# SVM
from sklearn.svm import SVC

svm = SVC()
svm.fit(x_train, y_train)
print('SVM score:', svm.score(x_test, y_test))
```

* 2번에서 생성한 표준화 데이터 사용 O

```python
# Random Forest
from sklearn.ensemble import RandomForestClassifier

rf = RandomForestClassifier(n_jobs=-1)
rf.fit(x_train_scaled, y_train)
print('Random Forest score (scaled):', rf.score(x_test_scaled, y_test))
```

```python
# XGBoost
from xgboost import XGBClassifier

xgb = XGBClassifier()
xgb.fit(x_train_scaled, y_train)
print('XGBoost score (scaled):', xgb.score(x_test_scaled, y_test))
```

```python
# SVM
from sklearn.svm import SVC

svm = SVC()
svm.fit(x_train_scaled, y_train)
print('SVM score (scaled):', svm.score(x_test_scaled, y_test))
```
