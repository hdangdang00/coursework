## 컴퓨터비전응용 4주차 과제

1. vscode를 이용하여 워크스테이션 서버에 접속한 후, 아래 위치의 이미지를 읽어와 채널을 나눠 출력해보세요.

```python
import cv2
image = cv2.imread('./img/chromakey.jpg')
b, g, r = cv2.split(image)

cv2.imshow('image', image)
cv2.imshow('b', b)
cv2.imshow('g', g)
cv2.imshow('r', r)
cv2.waitKey()
cv2.destroyAllWindows()
```

2. `inRange()` 함수를 써서 g채널의 밝기가 160부터 255까지의 픽셀을 255, 그렇지 않은 픽셀을 0으로 변환 출력하세요.

```python
import cv2
image = cv2.imread('./img/chromakey.jpg')
b, g, r = cv2.split(image)
chroma_image = cv2.inRange(g, 160, 255)

cv2.imshow('chroma_image', chroma_image)
cv2.waitKey()
cv2.destroyAllWindows()
```

3. `bitwise_not()`을 사용하여 2번 문제의 chroma_image를 반전해 출력하세요.

```python
import cv2
image = cv2.imread('./img/chromakey.jpg')
b, g, r = cv2.split(image)
chroma_image = cv2.inRange(g, 160, 255)
chroma_inverted = cv2.bitwise_not(chroma_image)

cv2.imshow('chroma_image', chroma_image)
cv2.imshow('chroma_inverted', chroma_inverted)
cv2.waitKey()
cv2.destroyAllWindows()
```

4. `bitwise_and()`를 사용하여 image의 사람 객체만 출력해 보세요. (배경을 마스킹)

```python
import cv2
image = cv2.imread('./img/chromakey.jpg')
b, g, r = cv2.split(image)
chroma_image = cv2.inRange(g, 160, 255)
chroma_inverted = cv2.bitwise_not(chroma_image)
filtered = cv2.bitwise_and(image, image, mask=chroma_inverted)
cv2.imshow('filtered', filtered)
cv2.waitKey()
cv2.destroyAllWindows()
```
