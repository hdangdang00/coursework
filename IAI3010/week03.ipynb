```python
import tkinter as tk

root = tk.Tk()
root.geometry("420x240")

items = tk.Listbox(root, height=3, exportselection=False)
items.pack(pady=10)
items.insert(tk.END, "원본", "흑백", "윤곽선")

def choose():
    selected = items.curselection()
    if selected:
        print(items.get(selected[0]))

tk.Button(root, text="선택 확인", command=choose).pack()

root.mainloop()
```

```python
import tkinter as tk

root = tk.Tk()
root.geometry("420x240")

def read_memo():
    content = memo.get("1.0", "end-1c")
    print(content)

memo = tk.Text(root, width=30, height=4)
memo.pack(pady=10)
memo.insert("1.0", "오늘의 실습 기록\n")
tk.Button(root, text="내용 읽기", command=read_memo).pack()

root.mainloop()
```

```python
import tkinter as tk

root = tk.Tk()
root.geometry("420x240")

group = tk.LabelFrame(root, text="처리기록", padx=12, pady=10)
group.pack(fill="both", expand=True, padx=20, pady=20)
memo = tk.Text(group, width=28, height=4)
memo.pack(fill="both", expand=True)

root.mainloop()
```

## 컴퓨터비전응용 3주차 과제

1. vscode를 워크스테이션 서버에 접속한 후, 아래와 같은 실행 결과가 나오게 코드를 완성하세요.

```python
import tkinter as tk # tkinter import

def greet():
    name = entry.get() # entry 객체 데이터 읽기
    if not name:
        result.config(text='이름을 먼저 입력하세요.')
        return
    # result 객체 텍스트 변경
    result.config(text=f"반갑습니다, {name}님!")

root = tk.Tk() # tk 윈도우 클래스 생성
root.title("나만의 이름표") # 윈도우 제목
root.geometry("420x240") # 윈도우 크기

entry = tk.Entry(root, width=20) # 글자 입력칸 생성
entry.pack(pady=8)
# 버튼 생성 → 함수 호출
tk.Button(root, text="인사하기", command=greet).pack()
result = tk.Label(root, text="") # 라벨 생성
result.pack(pady=8)
root.mainloop() # 메인루프
```

2. vscode를 워크스테이션 서버에 접속한 후, 아래와 같은 실행 결과가 나오게 코드를 완성하세요.

```python
import tkinter as tk

def add_record():
    text = entry.get()
    if not text:
        return
    items.insert(tk.END, text) # Listbox에 넣기
    memo.insert(tk.END, text + '\n') # Text에 넣기
    entry.delete(0, tk.END) # 다 지우기 = 리셋

root = tk.Tk()
root.title("영상 처리 기록장")
root.geometry("420x240")

entry = tk.Entry(root, width=24)
entry.pack(pady=6)

tk.Button(root, text="추가", command=add_record).pack()
items = tk.Listbox(root, height=4)
items.pack()
memo = tk.Text(root, width=30, height=4)
memo.pack(pady=6)
root.mainloop()
```
