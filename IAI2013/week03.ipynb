```python
# 1. 데이터셋 살펴보기 : FC_new.csv

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
import keras

# Feature : v(speed) = v / v_max     a(acceleration) = a * 5     sin(gradient)
# Target : FC_msec
FC = pd.read_csv('FC_new.csv')

print(FC.info())  
print(FC.describe().round(2))
feature = FC[['v','a','sin']].to_numpy()
target = FC['FC_msec'].to_numpy()

train_input, test_input, train_target, test_target = train_test_split(feature, target, test_size=0.2, random_state=42) 
val_input, test_input, val_target, test_target = train_test_split(test_input, test_target, test_size=0.5, random_state=42) 
print(train_input.shape, val_input.shape, test_input.shape)   
print(train_target.shape, val_target.shape, test_target.shape)   


# 2. 신경망 구성
model = keras.Sequential()
model.add(keras.layers.Dense(100, activation='relu'))
model.add(keras.layers.Dropout(0.3))
model.add(keras.layers.Dense(100, activation='relu'))
model.add(keras.layers.Dropout(0.3))
model.add(keras.layers.Dense(1))



# 3. 학습 : compile – fit – history (loss plot)
model.compile(optimizer='adam', loss='mse')
history = model.fit(train_input, train_target, epochs=10, validation_data=(val_input, val_target))
model.summary()
model.save('FC_reg1.keras')
plt.plot(history.history['loss']) 
plt.plot(history.history['val_loss'])
plt.xlabel('epoch')
plt.ylabel('loss')
plt.legend(['train', 'val'])
plt.show()


# 4. 추론(Inference) : predict for random 10 data
np.random.seed(42)
num = 10
ind = np.random.choice(len(test_input), num)
y = model.predict(test_input[ind]).round(1)
y = y.reshape(-1)
t = test_target[ind].round(1)
print(ind) 
print(y)
print(t)


# 5. 모델 평가 : random 10 data
Diff = t - y
N = len(t)
MSE  = sum(Diff**2) / N
RMSE = np.sqrt(MSE)
MAE  = sum(np.abs(Diff)) / N
MAPE = 100 * np.mean(np.abs(Diff)) / np.mean(np.abs(t))   # t=0 인 경우 고려
Error = np.array([MSE, RMSE, MAE, MAPE]).round(1)
print('[MSE, RMSE, MAE, MAPE]')
print(Error)


# 6. Regression graph for test dadatset
y = model.predict(test_input).reshape(-1)
t = test_target
Diff = t - y
N = len(t)
MSE  = sum(Diff**2) / N
RMSE = np.sqrt(MSE)
MAE  = sum(np.abs(Diff)) / N
MAPE = 100 * np.mean(np.abs(Diff)) / np.mean(np.abs(t))   # t=0 인 경우 고려
Error = np.array([MSE, RMSE, MAE, MAPE]).round(2)
print(Error)

plt.figure(figsize=(8, 6))
plt.scatter(t, y, alpha=0.7)
plt.plot([min(t), max(t)], [min(t), max(t)], 'r--')   # [x1, x2] ~ [y1, y2]
plt.xlabel('Actual Values (t)')
plt.ylabel('Predicted Values (y)')
plt.title('Actual vs. Predicted Values for Test Dataset')
plt.text(0, 42, f'MSE RMSE MAE MAPE = {Error}')
plt.show()
```

```python
# 1. 데이터셋 살펴보기 : FC_new.csv
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
import keras

FC = pd.read_csv('FC_new.csv')
feature = FC[['v','a','sin']].to_numpy()
target = FC['FC_msec'].to_numpy()
train_input, test_input, train_target, test_target = train_test_split(feature, target, test_size=0.2, random_state=42) 
val_input, test_input, val_target, test_target = train_test_split(test_input, test_target, test_size=0.5, random_state=42) 


# 2. 신경망 구성
model = keras.Sequential()
model.add(keras.layers.Dense(100, activation='relu'))
model.add(keras.layers.Dropout(0.3))
model.add(keras.layers.Dense(100, activation='relu'))
model.add(keras.layers.Dropout(0.3))
model.add(keras.layers.Dense(1))


# 3. 학습 : compile – fit
model.compile(optimizer='adam', loss='mse')
history = model.fit(train_input, train_target, epochs=10, validation_data=(val_input, val_target))
model.summary()
model.save('FC_reg1.keras')


# 6. Regression graph for test dataset
y = model.predict(test_input).reshape(-1)
t = test_target
Diff = t - y
N = len(t)
MSE  = sum(Diff**2) / N
RMSE = np.sqrt(MSE)
MAE  = sum(np.abs(Diff)) / N
MAPE = 100 * np.mean(np.abs(Diff)) / np.mean(np.abs(t))
R_squared = 1 - sum(Diff**2) / sum((t-np.mean(t))**2)
PC = np.corrcoef(t, y)[0, 1]
Error = np.array([MSE, RMSE, MAE, MAPE, R_squared, PC]).round(2)
print(Error)

plt.figure(figsize=(8, 6))
plt.scatter(t, y, alpha=0.7)
plt.plot([min(t), max(t)], [min(t), max(t)], 'r--')
plt.xlabel('Actual Values (t)')
plt.ylabel('Predicted Values (y)')
plt.title('Actual vs. Predicted Values for Test Dataset')
plt.text(0, 42, f'MSE RMSE MAE MAPE R^2 PC = {Error}')
plt.show()
```
